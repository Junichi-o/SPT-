```python
def alpha_T(a, t0, r):
    return (t0 * a**r) / (1 + t0 * a**r)  # 飽和型で常に < 1
```

```python
def mu_beyond(a, beta, s, t0, r):
    return (1 + 0.5 * alpha_T(a, t0, r)) / (1 + beta * a**(-s))
```

```python
def growth_eq_beyond(ln_a, y, beta, s, k0, p, b0, t0, r):
    a = np.exp(ln_a)
    D, dD_dlna = y
    d2D = - (2 + dlnH_dlna(a) + alpha_M(a, beta, s) + Gamma(a, k0, p, b0)) * dD_dlna \
          + 1.5 * mu_beyond(a, beta, s, t0, r) * Omega_m(a) * D
    return [dD_dlna, d2D]
```

```python
def predict_fs8_beyond(z_array, beta, s, k0, p, b0, t0, r):
    fs8_list = []
    a_ini = 1e-2
    y0 = [a_ini, 1.0]

    # D(0) を取得
    sol0 = solve_ivp(
        lambda ln_a, y: growth_eq_beyond(ln_a, y, beta, s, k0, p, b0, t0, r),
        (np.log(a_ini), 0.0), y0, method='RK45', rtol=1e-5, atol=1e-8
    )
    D0 = sol0.y[0, -1]

    for z in z_array:
        a = 1 / (1 + z)
        ln_a_span = (np.log(a_ini), np.log(a))

        sol = solve_ivp(
            lambda ln_a, y: growth_eq_beyond(ln_a, y, beta, s, k0, p, b0, t0, r),
            ln_a_span, y0, method='RK45', rtol=1e-5, atol=1e-8
        )

        if not sol.success or np.any(np.isnan(sol.y)):
            fs8_list.append(np.nan)
            continue

        D = sol.y[0, -1] / D0
        dD_dlna = sol.y[1, -1] / D0
        f = dD_dlna / D
        sigma8 = 0.8
        fs8 = f * D * sigma8
        fs8_list.append(fs8)

    return np.array(fs8_list)
```

```python
# Hubbleパラメータ（例：ΛCDM）
def H_z(z, H0=70.0, Omega_m0=0.3, Omega_L0=0.7):
    return H0 * np.sqrt(Omega_m0 * (1 + z)**3 + Omega_L0)

# H(a)
def H_a(a, H0=70.0, Omega_m0=0.3, Omega_L0=0.7):
    return H0 * np.sqrt(Omega_m0 / a**3 + Omega_L0)

# dlnH/dlna
def dlnH_dlna(a, Omega_m0=0.3, Omega_L0=0.7):
    E = np.sqrt(Omega_m0 / a**3 + Omega_L0)
    dE_dlna = -1.5 * Omega_m0 / a**3 / E
    return dE_dlna / E
```

```python
def Omega_m(a, Omega_m0=0.3, Omega_L0=0.7):
    return Omega_m0 / (Omega_m0 + Omega_L0 * a**3)

def alpha_M(a, beta, s):
    return beta * a**s

def Gamma(a, k0, p, b0):
    return b0 * (a / k0)**p
```

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# 定数
Omega_m0 = 0.3
Omega_L0 = 0.7
H0 = 70.0

# α_T(a) パラメータ
t0 = 0.03
r = 1.2
beta = 0.1
s = 2.0

# H(a)
def H(a):
    return H0 * np.sqrt(Omega_m0 * a**-3 + Omega_L0)

def dlnH_dlna(a):
    return -1.5 * Omega_m0 * a**-3 / (Omega_m0 * a**-3 + Omega_L0)

# α_T(a) と μ(a)
def alpha_T(a):
    return (t0 * a**r) / (1 + t0 * a**r)

def mu_eff(a):
    return (1 + 0.5 * alpha_T(a)) / (1 + beta * a**-s)

def Omega_m(a):
    return Omega_m0 * a**-3 / (Omega_m0 * a**-3 + Omega_L0)

# D₁(a) の微分方程式（x = ln a）
def dD1_dx(x, y):
    a = np.exp(x)
    D, Dp = y
    dlnH = dlnH_dlna(a)
    mu = mu_eff(a)
    Om = Omega_m(a)
    d2 = - (2 + dlnH) * Dp + 1.5 * mu * Om * D
    return [Dp, d2]

# 初期条件（a = 1e-3）
x0 = np.log(1e-3)
x1 = np.log(1.0)
y0 = [1e-5, 1e-5]  # D₁ ≈ a, D₁' ≈ a

sol = solve_ivp(dD1_dx, [x0, x1], y0, t_eval=np.linspace(x0, x1, 200))
a_vals = np.exp(sol.t)
D1_vals = sol.y[0]
D1_vals /= D1_vals[-1]  # 正規化 D₁(a=1) = 1

# プロット
plt.plot(a_vals, D1_vals, label=r"$D_1(a)$ with $\alpha_T$")
plt.xlabel("Scale factor a")
plt.ylabel("Growth function D₁(a)")
plt.title("Linear Growth Function with α_T(a)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 成長率 f(a) = dlnD₁/dlna
f_vals = np.gradient(np.log(D1_vals), np.log(a_vals))

# ε(a) = dlnf/dlna + f²
dlnf_dlna = np.gradient(np.log(f_vals), np.log(a_vals))
epsilon_vals = dlnf_dlna + f_vals**2

# F₂ カーネル係数
c1 = (3/7) * (1 + (2/5) * epsilon_vals)
c2 = (1/2) * (1 + (2/5) * epsilon_vals)
c3 = (4/7) * (1 - (3/5) * epsilon_vals)

# プロット
plt.plot(a_vals, c1, label=r"$c_1(a)$")
plt.plot(a_vals, c2, label=r"$c_2(a)$")
plt.plot(a_vals, c3, label=r"$c_3(a)$")
plt.xlabel("Scale factor a")
plt.ylabel("F₂ kernel coefficients")
plt.title("Time Evolution of F₂ Kernel Coefficients with α_T(a)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
from scipy.integrate import quad
from scipy.interpolate import interp1d

# 線形パワースペクトル（近似）
def P_L(k):
    A = 1e4  # 正規化（任意）
    n_s = 0.96
    return A * k**n_s * np.exp(-k/0.2)

# F₂ カーネル（μ = 0.5 で代表値）
def F2_eff(a):
    idx = np.argmin(np.abs(a_vals - a))
    return c1[idx] + 0.5 * c2[idx] + 0.25 * c3[idx]

# P22(k)
def P22(k, a):
    F2 = F2_eff(a)
    integrand = lambda q: q**2 * (F2**2) * P_L(q) * P_L(np.abs(k - q))
    result, _ = quad(integrand, 1e-3, 1.0, limit=100)
    return result * 2 / (2 * np.pi**2)

# P13(k)
def P13(k, a):
    F3 = 5/126  # EdS 近似
    integrand = lambda q: q**2 * P_L(q)
    result, _ = quad(integrand, 1e-3, 1.0, limit=100)
    return 6 * P_L(k) * F3 * result / (2 * np.pi**2)

# 合計
def P_1loop(k, a):
    return P_L(k) + P22(k, a) + P13(k, a)

# プロット
k_vals = np.logspace(-2, 1, 50)
a_target = 1.0
P_vals = [P_1loop(k, a_target) for k in k_vals]

plt.loglog(k_vals, [P_L(k) for k in k_vals], label="Linear P(k)")
plt.loglog(k_vals, P_vals, label="1-loop P(k) with α_T")
plt.xlabel("k [h/Mpc]")
plt.ylabel("P(k)")
plt.title("1-loop Power Spectrum with α_T(a)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
from numpy.linalg import eigvals, det

# 基準パラメータ
r0 = 1.2
t0_0 = 0.03
delta_r = 0.05
delta_t0 = 0.005
epsilon = 0.5

# α_T(a) と μ(a) の定義（可変）
def alpha_T(a, r, t0): return (t0 * a**r) / (1 + t0 * a**r)
def mu_eff(a, r, t0): return 1 + epsilon * alpha_T(a, r, t0)

# D₁(a) の数値解（再利用可能）
def compute_fsigma8(r, t0):
    def dD1_dx(x, y):
        a = np.exp(x)
        D, Dp = y
        dlnH = dlnH_dlna(a)
        mu = mu_eff(a, r, t0)
        Om = Omega_m(a)
        d2 = - (2 + dlnH) * Dp + 1.5 * mu * Om * D
        return [Dp, d2]

    sol = solve_ivp(dD1_dx, [x0, x1], y0, t_eval=np.linspace(x0, x1, 200))
    a_vals = np.exp(sol.t)
    D1_vals = sol.y[0]
    D1_vals /= D1_vals[-1]
    f_vals = np.gradient(np.log(D1_vals), np.log(a_vals))
    fsigma8_vals = f_vals * D1_vals * sigma8_0
    z_vals = 1 / a_vals - 1
    return np.interp(z_obs, z_vals[::-1], fsigma8_vals[::-1])  # 観測点に補間

# 偏導関数（数値微分）
fs8_base = compute_fsigma8(r0, t0_0)
fs8_r_plus = compute_fsigma8(r0 + delta_r, t0_0)
fs8_r_minus = compute_fsigma8(r0 - delta_r, t0_0)
fs8_t0_plus = compute_fsigma8(r0, t0_0 + delta_t0)
fs8_t0_minus = compute_fsigma8(r0, t0_0 - delta_t0)

df_dr = (fs8_r_plus - fs8_r_minus) / (2 * delta_r)
df_dt0 = (fs8_t0_plus - fs8_t0_minus) / (2 * delta_t0)

# Fisher行列の構築
F11 = np.sum((df_dr**2) / fs8_err**2)
F22 = np.sum((df_dt0**2) / fs8_err**2)
F12 = np.sum((df_dr * df_dt0) / fs8_err**2)
F = np.array([[F11, F12], [F12, F22]])

# 固有値・行列式
eigenvals = eigvals(F)
fisher_det = det(F)

# 出力
print("Fisher Matrix:\n", F)
print("Eigenvalues:", eigenvals)
print("Determinant:", fisher_det)
```

```python
from numpy.linalg import eig, inv

# 固有値・固有ベクトル
eigenvals, eigenvecs = eig(F)

# 並び替え（大きい固有値を先に）
idx = np.argsort(eigenvals)[::-1]
eigenvals = eigenvals[idx]
eigenvecs = eigenvecs[:, idx]

# 主成分方向のベクトル
v1 = eigenvecs[:, 0]  # 識別可能方向
v2 = eigenvecs[:, 1]  # 識別不能方向

print("主成分方向（識別可能）: v1 =", v1)
print("主成分方向（識別不能）: v2 =", v2)

# 新しいパラメータ系への変換関数
def to_principal_coords(r, t0):
    return np.dot(inv(eigenvecs), np.array([r, t0]))

def from_principal_coords(theta_plus, theta_minus):
    return np.dot(eigenvecs, np.array([theta_plus, theta_minus]))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# パラメータ設定
n_observers = 36
angles = np.linspace(0, 2 * np.pi, n_observers, endpoint=False)
a = 0.75  # 現在のスケール因子
T = 0.4   # 輪廻周期

# 観測者の属性
np.random.seed(42)
phase_shifts = np.random.uniform(0, 2 * np.pi, size=n_observers)

# 存在関数（輪廻）を定義
def existence_function(a, phi_shift, T=0.4):
    return 0.5 * (1 + np.sign(np.sin(2 * np.pi * a / T + phi_shift)))

# 描画
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'polar': True})
ax.set_title("Mandala of Observer Reincarnation", fontsize=14, pad=20)

for i in range(n_observers):
    angle = angles[i]
    r = 1.0
    chi = existence_function(a, phase_shifts[i])
    color = 'crimson' if chi == 1 else 'gray'
    marker = 'o' if chi == 1 else 'x'
    ax.plot([angle, angle], [0, r], color=color, linewidth=1.5, alpha=0.8)
    ax.plot(angle, r, marker=marker, color=color, markersize=8)

# 凡例
legend_elements = [
    plt.Line2D([0], [0], color='crimson', lw=2, label='Active Observer'),
    plt.Line2D([0], [0], color='gray', lw=1, label='Silent Observer'),
]
ax.legend(handles=legend_elements, loc='lower center', bbox_to_anchor=(0.5, -0.1), ncol=1)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go
from matplotlib import cm

# パラメータ設定
n_layers = 30              # 時間ステップ（球の層数）
n_theta = 24               # 緯度方向の分割数
n_phi = 48                 # 経度方向の分割数
omega = 4 * np.pi          # 螺旋のねじれ速度
r_min, r_max = 0.5, 2.0    # 球の最小・最大半径

# 球面座標を生成
theta_vals = np.linspace(0, np.pi, n_theta)
phi_vals = np.linspace(0, 2 * np.pi, n_phi)
theta, phi = np.meshgrid(theta_vals, phi_vals)

# データ格納
x_vals, y_vals, z_vals, colors = [], [], [], []

# カラーマップ
colormap = cm.get_cmap('plasma')

# 時間ステップごとに球殻を生成し、螺旋的にねじる
for i, t in enumerate(np.linspace(0, 1, n_layers)):
    r = r_min + (r_max - r_min) * t  # 時間に応じて半径が拡大
    twist = omega * t                # 螺旋のねじれ量

    # 球面座標をデカルト座標に変換（ねじれを加える）
    x = r * np.sin(theta) * np.cos(phi + twist)
    y = r * np.sin(theta) * np.sin(phi + twist)
    z = r * np.cos(theta)

    x_vals.extend(x.flatten())
    y_vals.extend(y.flatten())
    z_vals.extend(z.flatten())

    # 色は時間に応じて変化
    c = colormap(t)
    rgb = f'rgb({int(c[0]*255)}, {int(c[1]*255)}, {int(c[2]*255)})'
    colors.extend([rgb] * x.size)

# 可視化
fig = go.Figure(data=[go.Scatter3d(
    x=x_vals,
    y=y_vals,
    z=z_vals,
    mode='markers',
    marker=dict(
        size=2.5,
        color=colors,
        opacity=0.85
    )
)])

fig.update_layout(
    title='曼荼羅の時間的進化と螺旋構造の融合',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='data',
        camera=dict(eye=dict(x=1.6, y=1.6, z=1.2))
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()
```
